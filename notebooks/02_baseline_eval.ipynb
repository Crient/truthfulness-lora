```python
import socket
import torch

print("Notebook host:", socket.gethostname())
print("CUDA available:", torch.cuda.is_available())
print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "none")
print(torch.cuda.memory_summary() if torch.cuda.is_available() else "")
```

```python
import os
import sys
import subprocess
from pathlib import Path

PROJECT = Path(os.environ.get("TRUTHFULNESS_PROJECT", Path.cwd()))
os.environ["HF_HOME"] = str(PROJECT / "hf_cache")

# Start with 20 questions to confirm the model and evaluation task work.
LIMIT = 20
OUT = PROJECT / "results" / "baseline_smoke"
OUT.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable, "-m", "lm_eval", "run",
    "--model", "hf",
    "--model_args", "pretrained=Qwen/Qwen3-1.7B,dtype=float16",
    "--tasks", "truthfulqa_mc1,truthfulqa_mc2",
    "--device", "cuda:0",
    "--batch_size", "1",
    "--num_fewshot", "0",
    "--output_path", str(OUT),
    "--log_samples",
]

if LIMIT is not None:
    cmd += ["--limit", str(LIMIT)]

print("Running baseline smoke check...")
subprocess.run(cmd, check=True)
```
